```python
# Global scope
# x = 10

def outer_function():
    # Enclosing scope
    x = 20
    
    def inner_function():
        # Local scope
        x = 30
        print("Accessing from inner_function:", x)
    
    inner_function()
    print("Accessing from outer_function:", x)

outer_function()
print("Accessing from global scope:", x)
```
